## Handling Null Values

```python
df[["director", "cast", "country"]] = df[["director", "cast", "country"]].fillna("")
df["rating"] =  df["rating"].fillna(df["rating"].mode()[0])
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df['date_added']=df['date_added'].ffill()
```

```python
df.isnull().sum()
```

```python
movie_duration = df[df["type"] == "Movie"]["duration"].mode()[0]
df.loc[(df["type"] == "Movie") & (df["duration"].isnull()), "duration"] = movie_duration

tv_duration = df[df["type"] == "TV Show"]["duration"].mode()[0]
df.loc[(df["type"] == "TV Show") & (df["duration"].isnull()), "duration"] = tv_duration
```

## Data Visualization

## Rating Distribution by Year

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10,6))

sns.countplot(
    y="rating",
    data=df,
    order=df["rating"].value_counts().index
)

plt.title("Rating Distribution")
plt.xlabel("Count")
plt.ylabel("Rating")
plt.show()
```

## Content Released by Year

```python
release_counts = df["release_year"].value_counts().sort_index()

plt.figure(figsize=(12,5))
plt.plot(
    release_counts.index,
    release_counts.values,
    marker='o'
)
plt.title("Content Released by Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.grid(True)
plt.show()
```

```python
df["duration_time"] = df["duration"].str.split().str[0].astype(int)

df["Movie_time"] = 0
df["TV_Show_time"] = 0

df.loc[df["type"]== "Movie", "Movie_time"] = df.loc[df["type"]== "Movie", "duration_time"]
df.loc[df["type"]!= "Movie", "TV_Show_time"] = df.loc[df["type"]!= "Movie", "duration_time"]
```

```python
plt.figure(figsize=(12,5))

sns.kdeplot(
    df.loc[df["Movie_time"] != 0, "Movie_time"],
    label="Movie Time",
    fill=True
)
sns.kdeplot(
    df.loc[df["TV_Show_time"] != 0, "TV_Show_time"],
    label="TV Show Seasons",
    fill=True
)
plt.title("Distribution of Movie Duration and TV Show Seasons")
plt.xlabel("Duration")
plt.ylabel("Density")
plt.legend()
plt.show()
```

## Most frequent Actors 

```python
actors = df["cast"].str.split(", ").explode()
print(actors.value_counts().head(15))
```

```python
actors = df["cast"].dropna().str.split(", ").explode()

actors = actors[
    ~actors.isin(["Unknown", "Not Available", "", "nan", "NaN", "None"])
]
top_10_actors = actors.value_counts().head(10)
```

```python
print(top_10_actors)
```

```python
plt.figure(figsize=(10,6))

sns.barplot(
    x=top_10_actors.values,
    y=top_10_actors.index
)

plt.title("Top 10 Most Frequent Actors")
plt.xlabel("Number of Titles")
plt.ylabel("Actor")
plt.show()
```
